```python
number = int(input("enter some positive number greater than 2:"))
print("number", number, sep='-->>')

div_num = 2
while div_num<number:
    if number%div_num == 0: 
        print(number, "is NOT a PRIME Number.")
        break
    div_num += 1
else:
    print(number, "is a PRIME Number")
    
print("I am done.")
```
